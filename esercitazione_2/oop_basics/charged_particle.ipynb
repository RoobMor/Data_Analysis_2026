```python
import math
import vector
from particle import particle
```

```python
class charged_particle(particle):
    def __init__(self, px, py, pz, m=-1, E=-1, charge=None):
        
        self.p = math.sqrt(px**2 + py**2 + pz**2)
        
        if charge is None:
            print("Warning! Charge not specified! Given default: charge = 0")
            self.charge = 0
        else:
            self.charge = charge

        if m != -1:
            if m == 0:
                print("Warning! Massless particle!")
                # sarebbe carino definire a questo punto una classe per le massless, e passarla per ereditarietà, al pari di particle.
            if m < 0:
                print("Warning! Negative mass spotted!")
            # self.m = m
            # self.E = math.sqrt(self.p**2 + m**2)
    
            # il metodo super() permette di agganciarsi alla superclasse (ergo, il padre) e sfruttarne l'ereditarietà.
            # da leggersi come: dalla classe superiore, prendi il seguente metodo (nel nostro caso: __init__).
            # in questo modo definiamo in un colpo solo tutti gli attributi che self riceve nell'init di particle.
            super().__init__(px, py, pz, m)

        if E != -1:
            self.E = E
            
            if (E**2 - self.p**2) < 0:
                print("Warning! E > p + m, unphysical particle!")
            
            if (m != -1) and (m != self.m):
                print("Warning! E != p + m, mass shell violated!")

        if (m == -1):
            if (E == -1):
                print("Warning! Missing parameters!")
            else:
                self.m = math.sqrt(E**2 - self.p**2)
            
            super().__init__(px, py, pz, m)
            self.p4 = vector.obj(px=px, py=py, pz=pz, E=self.E)

    def charged_particle_p4(lorentz_p4, charge):
        return charged_particle(px = lorentz_p4.px, py = lorentz_p4.py, pz = lorentz_p4.pz, E = lorentz_p4.E, charge = charge)

    # ridefinisco la somma tra oggetti della stessa classe
    def __add__(self, other):
        # nel caso in cui venissero sommati due tipi/classi diversi
        if not isinstance(other, particle):
            return NotImplemented
        else:
            # è banale, solo perchè ho definito p4 come vector.obj(...)
            sum_p4 = self.p4 + other.p4
            sum_charge = self.charge + other.charge
            
            # altrimenti avrei dovuto fare:
            # sum_x = self.px + other.px
            # sum_y = self.py + other.py
            # sum_z = self.pz + other.pz
            # sum_E = self.E  + other.E
            # sum_charge = self.charge + other.charge

            return self.charged_particle_p4(lorentz_p4 = sum_p4, charge = sum_charge)

    # ridefinisco il significato di stampa di un oggetto della classe
    def __str__(self):
        return (f"\nCharacteristics of the particle:\n px = {self.px}\n py = {self.py}\n pz = {self.pz}\n m  = {self.m}\n Q  = {self.charge}\n \nMomentum-Energy 4-vector:\n p4 = {self.p4}\n \nCinematic variables:\n Gamma = {self.gamma}\n Beta  = {self.beta}")
```
