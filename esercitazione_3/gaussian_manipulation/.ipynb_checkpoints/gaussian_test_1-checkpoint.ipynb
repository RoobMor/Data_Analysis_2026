```python
import numpy as np
from scipy.stats import norm
from scipy.optimize import curve_fit
import matplotlib.pyplot as plt
```

```python
x_data = np.arange(-20,20,0.0001)
y_data = norm.pdf(x_data, 0, 1)

fig, ax = plt.subplots()
ax.set_title(rf"Normal distributions with $\mu = 0$")
ax.set_xlabel("x")
ax.set_ylabel("y")

ax.plot(x_data, y_data, label=rf"$\sigma = 1$")

for s in range(4):
    y = norm.pdf(x_data, 0, s+2)
    ax.plot(x_data, y, label=rf"$\sigma = {s+2}$")

ax.grid(alpha=0.3)
ax.legend()
plt.show
```

```python
x_data = np.arange(-20,20,0.0001)
y_data = norm.pdf(x_data, 0, 1)

fig, ax = plt.subplots()
ax.set_title(rf"Normal distributions with $\sigma = 0$")
ax.set_xlabel("x")
ax.set_ylabel("y")

ax.plot(x_data, y_data, label=rf"$\mu = 0$")

for m in range(4):
    y = norm.pdf(x_data, 3*m+2, 1)
    ax.plot(x_data, y, label=rf"$\mu = {3*m+2}$")

ax.grid(alpha=0.3)
ax.legend()
plt.show
```

```python
delta = (max(x_data) - min(x_data))/len(x_data)
area = np.trapezoid(y_data, dx = delta)
print(f"area = {area}")

data = np.random.normal(loc = 0.5, scale = 3.0, size = 5000)
mean, std = norm.fit(data)
print("mean =", mean)
print("std =", std)
nbins = 75

plt.hist(data, nbins, density=True)

x_min, x_max = plt.xlim()
x = np.linspace(x_min, x_max, 100)
y = norm.pdf(x, mean, std)
plt.plot(x, y)

plt.grid(alpha = 0.3)
plt.show
```

```python
fig, ax = plt.subplots()
n, bins, patches = ax.hist(data, nbins, density=True, facecolor = 'grey', alpha = 0.5, label='before');
centers = (0.5*(bins[1:]+bins[:-1]))
pars, cov = curve_fit(lambda data, mu, sig : norm.pdf(data, loc=mu, scale=sig), centers, n, p0=[0,1])

ax.plot(centers, norm.pdf(centers,*pars), 'k--',linewidth = 2, label='fit before')
ax.set_title(r'$\mu={:.4f}\pm{:.4f}$,     $\sigma={:.4f}\pm{:.4f}$'.format(pars[0],np.sqrt(cov[0,0]), pars[1], np.sqrt(cov[1,1 ])))

plt.plot(x, y)
plt.show()
```
